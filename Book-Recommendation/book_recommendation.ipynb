```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

data = pd.read_csv("book.csv")
data["description"] = data["description"].fillna("")
print(data)
```

```python
data=data.drop(columns=["Unnamed: 0","book_id","language_code","image_url"])
data.info()
```

```python
data.info()
```

```python
books=pd.DataFrame(data)
print(books)
```

```python
def get_content_based_recommendations(book_title, books, n_recommendations=5):
    tfidf = TfidfVectorizer(stop_words='english')
    tfidf_matrix = tfidf.fit_transform(books['description'])
    cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
    cosine_sim_df = pd.DataFrame(cosine_sim, index=books['title'], columns=books['title'])
    if book_title not in cosine_sim_df.columns:
        raise ValueError(f"Book title not found: {book_title}")
    sim_scores = cosine_sim_df[book_title].sort_values(ascending=False)
    similar_books = sim_scores[sim_scores.index != book_title].head(n_recommendations)
    return similar_books
```

```python
books_to_recommend= input("enter the book name")
recommendations=get_content_based_recommendations(books_to_recommend,books)
print(f"recommendations for {books_to_recommend}:")
print(recommendations)
```
