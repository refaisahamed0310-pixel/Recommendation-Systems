```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
data=pd.read_csv('MoviesOnStreamingPlatforms_updated.csv')
print(data)
```

```python
data.columns
```

```python
data=data.drop(columns=['Unnamed: 0', 'Year', 'Age',  'Rotten Tomatoes',
       'Netflix', 'Hulu', 'Prime Video', 'Disney+', 'Type', 'Directors',
       'Genres', 'Country', 'Language', 'Runtime'],axis=1)
```

```python
ratings_df = pd.DataFrame(data)
ratings_df.set_index('name', inplace=True)
print(ratings_df)
```

```python
similarity_matrix = cosine_similarity(ratings_df.fillna(0))
similarity_df = pd.DataFrame(similarity_matrix, index=ratings_df.index, columns=ratings_df.index)
print(similarity_df)
```

```python
# Earlier user-based recommendation experiment retained for reference.
# The dataset is organized as movies (rows) and streaming platforms (columns),
# so the item-item recommendation function in the next cell is used for movie suggestions.
```

```python
import pandas as pd

# Redefining the get_recommendations function for item-item similarity.
# 'item_name' here refers to a movie title, and we use the pre-calculated similarity_df.
def get_recommendations(item_name, similarity_df, n_recommendations=2):
    # Get the similarity scores for the given item (movie) from the similarity matrix
    # The 'item_name' here corresponds to the index (movie title) in similarity_df
    if item_name not in similarity_df.columns:
        raise ValueError(f'Movie title not found: {item_name}')
    item_similarities = similarity_df[item_name].sort_values(ascending=False)

    # Exclude the item itself from the recommendations
    # and get the top N most similar items
    recommendations = item_similarities.drop(item_name).nlargest(n_recommendations)
    return recommendations
user_to_recommend = 'Taimanin Asagi 3'
recommendations = get_recommendations(user_to_recommend, similarity_df)
print(f"Recommendations for {user_to_recommend}:")
print(recommendations)
```
