```python
import random
import pandas as pd

# 1. Define courses (replace User)
courses = [
    "Java",
    "Python Full Stack",
    "AI with Data Science",
    "Data Science",
    "Machine Learning",
    "Web Development",
    "Android Development",
    "Cyber Security",
    "Cloud Computing",
    "DevOps"
]

# 2. Define trainers (replace categories)
trainers = [
    "Vijay",
    "Meena",
    "Arun",
    "Priya",
    "Karthik",
    "Suresh",
    "Divya",
    "Ramesh"
]

# 3. Create dataset
data = {
    'Course': courses
}

# 4. Assign random suitability ratings (0–5)
for trainer in trainers:
    data[trainer] = [random.randint(1, 5) for _ in range(len(courses))]

# 5. Convert to DataFrame & save
df = pd.DataFrame(data)
df.to_csv("course_trainer_data.csv", index=False)

print("Dataset 'course_trainer_data.csv' generated successfully!")
```

```python
import pandas as pd
data=pd.read_csv("course_trainer_data.csv")
print(data)
```

```python
rating_df=pd.DataFrame(data)
rating_df.set_index("Course",inplace=True)
print(rating_df)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
matrix=cosine_similarity(rating_df.fillna(0))
similarity=pd.DataFrame(matrix,index=rating_df.index,columns=rating_df.index)
print(similarity)
```

```python
def get_recommendations(user, rating_df, similarity, n_recommendations=6):
    user_ratings = rating_df.loc[user]
    similar_users = similarity[user].sort_values(ascending=False)
    weighted_ratings = pd.Series(0.0, index=rating_df.columns, dtype=float)
    for similar_user, similarity_score in similar_users.items():
        if similar_user != user:
            similar_user_ratings = rating_df.loc[similar_user]
            weighted_ratings = weighted_ratings.add(similar_user_ratings * similarity_score, fill_value=0)

    recommendations =weighted_ratings[user_ratings == 0]
    return recommendations.nlargest(n_recommendations)


user_to_recommend = 'AI with Data Science'
recommendations = get_recommendations(user_to_recommend, rating_df, similarity)
print(f"Recommendations for {user_to_recommend}:")
print(recommendations)
```
