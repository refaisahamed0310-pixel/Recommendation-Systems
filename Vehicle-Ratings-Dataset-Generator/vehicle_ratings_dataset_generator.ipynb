```python
import random
random.seed(42)  # Make generated sample ratings reproducible.
import pandas as pd

# 1. Define vehicle categories (columns)
vehicle_categories = ["Maruti Suzuki",'Hyundai','Mahindra','Tata Motors','Toyota','Kia','Honda','Mercedes-Benz','BMW','Audi','Land Rover','Volvo','Lexus','Ford','Chevrolet','GMC','Ashok Leyland','Tesla']

# 2. Generate data for 1000 users
data = {
    'User': [f'User_{i+1}' for i in range(1000)]
}

# 3. Assign random ratings (0-5) for each vehicle type
for vehicle in vehicle_categories:
    data[vehicle] = [random.randint(0, 5) for _ in range(1000)]

# 4. Convert dictionary to DataFrame and save to CSV
df = pd.DataFrame(data)
df.to_csv('vehicle_data.csv', index=False)

print(f"Successfully generated 'vehicle_data.csv' with {len(df)} rows.")
```
